```python
import pandas as pd
import os
```

```python
DATA_DIR = "C:\\Users\\enkht\\OneDrive\\Pulpit\\Lung Cancer dataset EliLilly"
OUTPUT_DIR = "C:\\Users\\enkht\\OneDrive\\Pulpit\\Lung Cancer dataset EliLilly\\ready_for_bigquery\\"
```

```python
os.makedirs(OUTPUT_DIR,exist_ok = True)
```

```python
files_to_process = {
    "stg_demographics": "subjinfo.sas7bdat",
    "stg_disposition": "disposit.sas7bdat",
    "stg_adverse_events": "events.sas7bdat"
}
```

```python

#Creating parquet files with uppercase column names
for table_name, filename in files_to_process.items():
    file_path = os.path.join(DATA_DIR, filename)
    
    if os.path.exists(file_path):
        print(f"Processing domain: {filename}...")
        
        
        df = pd.read_sas(file_path, format='sas7bdat', encoding='latin-1')
        

        df.columns = [col.upper().strip() for col in df.columns]


        output_file = os.path.join(OUTPUT_DIR, f"{table_name}.parquet")
        df.to_parquet(output_file, index=False, engine='pyarrow')
```
