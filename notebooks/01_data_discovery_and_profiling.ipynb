```python
import pandas as pd
import os
```

```python
DATA_DIR = "C:\\Users\\enkht\\OneDrive\\Pulpit\\Lung Cancer dataset EliLilly"
```

```python
files_to_profile = {
    "Demographics" : "subjinfo.sas7bdat",
    "Disposition" : "disposit.sas7bdat",
    "Adverse_Events" : "events.sas7bdat"    
}
```

```python
for domain, filename in files_to_profile.items():
    file_path = os.path.join(DATA_DIR, filename)
    
    if not os.path.exists(file_path):
        print(f"File not found: {filename}")
        continue
    try:
        df = pd.read_sas(file_path, format = 'sas7bdat', encoding='latin-1', iterator=False)
        print(f"Domain: {domain} ({filename})")
        print(f"Table dimensions: {df.shape[0]} rows, {df.shape[1]} columns)")
        print(f"Column list:")
        print(list(df.columns))

              
    except Exception as e:
        print(f"Error while reading file {filename}: {str(e)}")
```
